```python
 #PROBLEM 1
### PART 1:-This function computes the Greatest Common Divisor (GCD) of two integers using the Euclidean Algorithm. The GCD is the largest number that divides both numbers without leaving a remainder.
def great(a,b):
    if b>a:
        a=b
        b=a
        while b!=0:
            r=a%b
            a=b
            b=r
    else:
        while b!=0:
            r=a%b
            a=b
            b=r
    return a
g=int(input("Enter a number: "))
h=int(input("Enter a number: "))
l=great(g,h)
print('Greatest Commmon Divisor is :',l)

```

```python
### Part 3
###Deﬁne a function called `euler_sum` that takes a single non-negative integer argument `n`, and returns an approximation to $e$ based on the ﬁrst `n` terms of the sum in Equation (1). Your function should return a ﬂoat.
###ou may assume that the input will be a non-negative integer, so you do not need to include error checking in your function.
import math

def euler_sum(terms):
    sum=0
    for k in range(terms):
        sum+=1/math.factorial(k)
    return sum
euler_sum(0)
```

```python
### Part 4
###Deﬁne a function called `euler_approx` that takes a single argument, a positive ﬂoat `epsilon`, and uses the sum in Equation (1) to obtain an approximation of $e$ that is within `epsilon` of the true value of $e$.
import math
def euler_approx(epsilon):
    s=0
    k=0
    while True:
        fact=1
        i=1
        while i<=k:
            fact*=i
            i+=1
        s=s+(1/fact)
        if abs(s-math.exp(1))<epsilon:
            break
        k=k+1
    return s
print(euler_approx(0.001))

```

```python
### Part 5
###For a value of interest $x$ and an approximation $\hat x$, we say the difference $\hat x - x$ is the **error** of $f(k)$.
import math
def print_euler_sum_error(n):
    for k in range(1,n+1):
        error=euler_sum(k)-math.exp(1)
        print(error)
def print_euler_lim_error(n):
    for k in range(1,n+1):
        error=euler_limit(k)-math.exp(1)
        print(error)
print_euler_sum_error(2)
print_euler_lim_error(2)

```

```python
### PROBLEM 3 :- TESTING PROPERTIES OF INTERGERS
# PART 1 :-
def is_power_of_2(n):
    while n>1:
        if n%2!=0:
            return False
        n=n//2
    return True
print(is_power_of_2(48))
```
